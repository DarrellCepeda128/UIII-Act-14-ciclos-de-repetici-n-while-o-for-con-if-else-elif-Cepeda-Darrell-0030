```python
# Problema Número 3
# Redacción: Diseñar un algoritmo interactivo que muestre la tabla de multiplicar
# de los números 1, 2, 3, 4 y 5, permitiendo al usuario definir hasta qué
# número desea multiplicar (límite).

print("==========================================")
print("   SISTEMA DE GENERACIÓN DE TABLAS IA   ")
print("   Alumno: Darrell Cepeda | Grupo: 2-H  ")
print("==========================================")

# Entrada de datos interactiva
limite = int(input("¿Hasta qué número quieres las tablas? (ej. 10): "))

# Inicialización de la variable para el número base (1 al 5)
numero_base = 1

# Bucle externo para controlar las tablas del 1 al 5 [4]
while numero_base <= 5:
    print(f"\n--- TABLA DEL {numero_base} ---")

    # Inicialización del multiplicador
    multiplicador = 1

    # Bucle interno para calcular cada multiplicación hasta el límite [5]
    while multiplicador <= limite:
        # Operación aritmética de producto [6]
        resultado = numero_base * multiplicador

        # Salida de resultados con formato [7]
        print(numero_base, "x", multiplicador, "=", resultado)

        # Incremento del multiplicador [8]
        multiplicador = multiplicador + 1

    # Incremento del número base para pasar a la siguiente tabla [8]
    numero_base = numero_base + 1

print("\n==========================================")
print("      PROCESO LÓGICO FINALIZADO         ")
print("==========================================")

print("¨Programa realizado por: Cepeda Darrell NL = 10")
```
